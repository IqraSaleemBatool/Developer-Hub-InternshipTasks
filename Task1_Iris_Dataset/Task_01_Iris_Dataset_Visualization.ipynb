## Introduction

The objective of this task is to explore and visualize the Iris dataset using Python. The dataset contains measurements of iris flowers and is widely used for data analysis and classification tasks.

In this task, we will perform data exploration and create visualizations to understand relationships between variables.



```python
# importing libraries for data visualization and analysis
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

## Dataset Description

The dataset consists of 150 observations(Rows) and 5 features(Columns):

* Sepal Length
* Sepal Width
* Petal Length
* Petal Width
* Species

These features help in identifying different species of iris flowers.


```python
# loading Iris data by using  built-in  datasetlibrary seaborn
df = sns.load_dataset('iris')
```

```python
# Displaying the shape of dataset as (rows, columns)
print("Shape of dataset:", df.shape)

# Displaying  column names
print("Columns:", df.columns)

# Displaying first 5 rows
print("First 5 rows:")
print(df.head())

# Summary statistics of numerical columns
print("Statistical Summary:")
print(df.describe())
```

## Data Cleaning and Preparation

The dataset was checked for missing values and inconsistencies. No missing values were found, so no major data cleaning was required.


```python
# Checking  if dataset contains any missing values
print("Missing values:\n", df.isnull().sum())
```

```python
# scatter plot for analzing realtionship between sepal length and width
sns.scatterplot(x='sepal_length', y='sepal_width', hue='species', data=df)
# adding title
plt.title("Sepal Length vs Sepal Width (by species)")
# Finally showing plot
plt.show()
```

```python
# Histogram for checking distribution of petal length
plt.hist(df['petal_length'])
# adding title
plt.title("Distribution of Petal Length")
# finally showing plot
plt.show()
```

```python
# box plot for detecting outliers in sepal length caross species
sns.boxplot(x='species', y='sepal_length', data=df)
# Adding tiltle for box plot
plt.title("Sepal Length Distribution by species")
# showing plot
plt.show()
```

```python
# Pairplot shows relationships between all variables
sns.pairplot(df, hue='species')
# title for plot
plt.title("Pair plot of dataset")
# showing plot
plt.show()
```

## Conclusion

* The dataset contains no missing values and well-structured.
* Petal length and width are strong indicators of species classification.
* Setosa species is clearly distinguishable from others.
* Visualizations helped in understanding feature relationships effectively.
